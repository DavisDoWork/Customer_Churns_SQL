## B. Data Cleaning

1. Finding the total number of customers



```python
%%sql
SELECT COUNT(DISTINCT(CustomerID)) As total_customer
FROM customers
```

2. Check for duplicate rows

```python
%%sql
SELECT CustomerID, Count(CustomerID) AS Count
FROM customers
GROUP BY CustomerID
HAVING COUNT(CustomerID) >1
```

3. Checking for null values

```python
%%sql 
SELECT 'Age' AS ColumnName, Count(*) AS NullCount
FROM customers
WHERE Age IS NULL
UNION
SELECT 'Gender' AS ColumnName, Count(*) AS NullCount
FROM customers
WHERE 'Gender' IS NULL
UNION
SELECT 'Tenure' AS ColumnName, Count(*) AS NullCount
FROM customers
WHERE 'Tenure' IS NULL
UNION
SELECT 'Usage Frequency' AS ColumnName, Count(*) AS NullCount
FROM customers
WHERE 'Usage Frequency' IS NULL
UNION
SELECT 'Support Calls' AS ColumnName, Count(*) AS NullCount
FROM customers
WHERE 'Support Calls' IS NULL
UNION
SELECT 'Payment Delay' AS ColumnName, Count(*) AS NullCount
FROM customers
WHERE 'Payment Delay' IS NULL
UNION
SELECT 'Subscription Type' AS ColumnName, Count(*) AS NullCount
FROM customers
WHERE 'Subscription Type' IS NULL
UNION
SELECT 'Contract Length' AS ColumnName, Count(*) AS NullCount
FROM customers
WHERE 'Contract Length' IS NULL
UNION
SELECT 'Subscription Type' AS ColumnName, Count(*) AS NullCount
FROM customers
WHERE 'Subscription Type' IS NULL
UNION
SELECT 'Total Spend' AS ColumnName, Count(*) AS NullCount
FROM customers
WHERE 'Total Spend' IS NULL
UNION
SELECT 'Last Interaction' AS ColumnName, Count(*) AS NullCount
FROM customers
WHERE 'Last Interaction' IS NULL
UNION
SELECT 'Churn' AS ColumnName, Count(*) AS NullCount
FROM customers
WHERE 'Churn' IS NULL

```

4. Creating a new column from an already existing “churn” column

```python
%%sql
ALTER TABLE customers
ADD COLUMN CustomerStatus VARCHAR(50);

UPDATE customers
SET CustomerStatus = CASE
    WHEN Churn = 1 THEN 'Churned'
    WHEN Churn = 0 THEN 'Stayed'
END;
```

## C. Data Exploration
# Answering business questions.




1. What is the overall customer churn rate?

```python
%%sql
SELECT
    COUNT(*) AS total_customers,
    SUM(CASE WHEN CustomerStatus = 'Churned' THEN 1 ELSE 0 END) AS churned_customers,
    ROUND(SUM(CASE WHEN CustomerStatus = 'Churned' THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) AS ChurnRate
FROM customers;
```

2. How does the churn rate vary based on the preferred login device?

```python
%%sql
SELECT 
    "Subscription Type",
    COUNT(*) AS TotalCustomers,
    SUM(CASE WHEN CustomerStatus = "Churned" THEN 1 ELSE 0 END) AS ChurnedCustomers,
    ROUND(SUM(CASE WHEN CustomerStatus = "Churned" THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) AS ChurnRate
FROM customers
GROUP BY "Subscription Type"
```

3.  Is there any correlation between age and customer churn? (There is no trend between AgeGroup)

```python
%%sql
ALTER TABLE customers
ADD COLUMN AgeGroup VARCHAR(10);
```

```python
%%sql
UPDATE customers
SET AgeGroup = CASE
    WHEN Age BETWEEN 18 AND 24 THEN '18-24'
    WHEN Age BETWEEN 25 AND 34 THEN '25-34'
    WHEN Age BETWEEN 35 AND 44 THEN '35-44'
    WHEN Age BETWEEN 45 AND 54 THEN '45-54'
    WHEN Age BETWEEN 55 AND 65 THEN '55-65'
    ELSE 'Other'
END;
```

```python
%%sql
SELECT
    AgeGroup,
    COUNT(*) AS TotalCustomer,
    SUM(CASE WHEN CustomerStatus = 'Churned' THEN 1 ELSE 0 END) AS CustomerChurn,
    ROUND(SUM(CASE WHEN CustomerStatus = 'Churned' THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) AS ChurnRate

FROM customers
GROUP BY AgeGroup
```

4. What is the typical tenure for churned customers? (Stay for longer tends to have lower churn rate)

```python
%%sql
SELECT MIN(Tenure), MAX(Tenure)
FROM customers
```

```python
%%sql
ALTER TABLE customers
ADD COLUMN TenureRange VARCHAR(10);
```

```python
%%sql
UPDATE customers
SET TenureRange = CASE
    WHEN Tenure BETWEEN 0 AND 12 THEN '1 year'
    WHEN Tenure BETWEEN 13 AND 24 THEN '2 year'
    WHEN Tenure BETWEEN 25 AND 36 THEN '3 year'
    WHEN Tenure BETWEEN 37 AND 48 THEN '4 year'
    WHEN Tenure BETWEEN 49 AND 60 THEN '5 year'
END
```

```python
%%sql
SELECT
    TenureRange,
    COUNT(*) AS Total,
    SUM(CASE WHEN CustomerStatus = 'Churned' THEN 1 ELSE 0 END) AS ChurnedCustomers,
    ROUND(SUM(CASE WHEN CustomerStatus = 'Churned' THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) AS ChurnRate
FROM customers
GROUP BY TenureRange
ORDER BY MIN(Tenure);
```

5. Is there any difference in churn rate between male and female customers?

```python
%%sql
SELECT 
    Gender,
    COUNT(*) AS TotalCustomer,
    SUM(Churn) AS CustomerChurn,
    ROUND(SUM(CASE WHEN CustomerStatus = 'Churned' THEN 1 ELSE 0 END) * 100.0 /COUNT(*), 2) AS Churnrate
FROM customers
GROUP BY Gender
```

6. Is there any difference in churn rate and Support Calls?

```python
%%sql
ALTER TABLE customers
ADD COLUMN SupportCallGroup VARCHAR(10);
```

```python
%%sql
UPDATE customers
SET SupportCallGroup = CASE
    WHEN "Support Calls" = 0 THEN '0'
    WHEN "Support Calls" BETWEEN 1 AND 3 THEN '1-3'
    WHEN "Support Calls" BETWEEN 4 AND 6 THEN '4-6'
    WHEN "Support Calls" BETWEEN 7 AND 9 THEN '7-9'
    ELSE '10+'
END;
```

```python
%%sql
SELECT
    SupportCallGroup,
    COUNT(*) AS Total,
    ROUND(SUM(CASE WHEN CustomerStatus = 'Churned' THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) AS ChurnRate
FROM customers
GROUP BY SupportCallGroup
ORDER BY MIN("Support Calls");
```

7. Gender by subscription type, to see whether the gender gap holds for every plan:

```python
%%sql
SELECT Gender, "Subscription Type", COUNT(*) AS Total,
    ROUND(SUM(CASE WHEN CustomerStatus = 'Churned' THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) AS ChurnRate
FROM customers
GROUP BY Gender, "Subscription Type"
ORDER BY Gender, "Subscription Type";
```

8. Spending quartiles, which uses a window function to split customers into four equal groups:

```python
%%sql
WITH quartiles AS (
    SELECT CustomerStatus, "Total Spend",
        NTILE(4) OVER (ORDER BY "Total Spend") AS SpendQuartile
    FROM customers
)
SELECT SpendQuartile,
    MIN("Total Spend") AS MinSpend,
    MAX("Total Spend") AS MaxSpend,
    COUNT(*) AS Total,
    ROUND(SUM(CASE WHEN CustomerStatus = 'Churned' THEN 1 ELSE 0 END) * 100.0 / COUNT(*), 2) AS ChurnRate
FROM quartiles
GROUP BY SpendQuartile
ORDER BY SpendQuartile;
```
